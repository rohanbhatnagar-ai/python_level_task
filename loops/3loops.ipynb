```python
# Q21. Count down from 10 to 0.
for i in range(10, -1, -1):
    print(i)
```

```python
# Q22. Print the first N natural numbers and their squares.
n = int(input("enter a number:"))

for i in range(1, n+1):
    print(i, "square=", i*i)
```

```python
# Q23. Find the average of numbers from 1 to N.
n = int(input("Enter a number: "))

total = 0

for i in range(1, n + 1):
    total = total + i

average = total / n

print("Sum =", total)
print("Average =", average)
```

```python
# Q24. Print all numbers between A and B (inclusive)
a = int(input("enter a:"))
b = int(input("enter b:"))

for i in range(a, b + 1):
    print(i)
```

```python
#Q25. Find the sum of numbers between A and B.
a = int(input("Enter A: "))
b = int(input("Enter B: "))

total = 0

for i in range(a, b + 1):
    total = total + i

print("Sum =", total)
```

```python
# Q26. Count how many numbers between 1 and N are multiples of 2 or 5.
n = int(input("Enter N: "))

count = 0

for i in range(1, n + 1):
    if i % 2 == 0 or i % 5 == 0:
        count = count + 1

print("Count =", count)
```

```python
#Q27. Print the first 10 multiples of a user-given number.
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
# Q28. Display each character of a word on a new line using a for loop.
word = input("enter a word:")

for ch in word:
    print(ch)
```

```python
# Q29. Count the characters in a string without using len().
text = input("Enter a string: ")

count = 0

for ch in text:
    count = count + 1

print("Total characters =", count)
```

```python
# Q30. Count vowels in a given string using a for loop.
text = input("enter a string:")
count = 0

for ch in text:
    if ch.lower() in "aeiou":
        count = count+1

```
