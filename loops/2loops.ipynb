```python
# Q13. Count how many numbers from 1 to N are divisible by 4.
n = int(input("enter the n:"))
count = 0
for i in range(1, n+1):
    if i%4 ==0:
        count += 1
print("Count:", count)
```

```python
# Q19. Find the sum of numbers divisible by 7 from 1 to N
n = int(input("enter the nn:"))
total = 0

for i in range(1, n+1):
    if i% 7 ==0:
        
        total = total + i

print("sum:", total)
```
