```python
# Q31. Count consonants in a given string using a for loop.
text = "Hello World"
count = 0

for ch in text:
    if ch.isalpha() and ch.lower() not in "aeiou":
        count +=1

print("Consonants:", count)
```

```python
# Q32. Count uppercase and lowercase letters in a string.

text = "Hello PYTHON"
upper = 0
lower = 0

for ch in text:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1

print("Uppercase:", upper)
print("Lowercase:", lower)
```

```python
# Q33. Find the sum of all items in a list without using sum().

numbers = [10, 20, 30, 40, 50]
total = 0

for num in numbers:
    total = total + num

print("Sum:", total)
```

```python
# Q34. Count even and odd values in a list.

numbers = [10, 15, 20, 25, 30, 35]

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
# Q35. Find the largest number in a list without using max().

numbers = [12, 45, 7, 89, 23]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest:", largest)
```

```python
# Q36. Find the smallest number in a list without using min().

numbers = [12, 45, 7, 89, 23]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print("Smallest:", smallest)
```

```python
# Q37. Print each item of a list with its index using a loop.


numbers = ["Python", "Java", "SQL", "ML"]

for i in range(len(numbers)):
    print(i, numbers[i])
```

```python
# Q38. Create a new list containing the squares of the original list values.

numbers = [2, 3, 4, 5]
squares = []

for num in numbers:
    squares.append(num * num)

print(squares)
```

```python
# Q39. Count how many times a given value appears in a list.

numbers = [10, 20, 10, 30, 10, 40]

target = 10
count = 0

for num in numbers:
    if num == target:
        count += 1

print("Occurrences:", count)
```
